# Project 3: Data Cleaning and Preparation

**Dataset:** IBM Telco Customer Churn Dataset

**Objective:** Clean the dataset by handling missing values, duplicates, incorrect data types, and inconsistent values to prepare it for Exploratory Data Analysis and Machine Learning.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe(include='all')
```

```python
df.columns
```

```python
(df == " ").sum()
```

```python
df.replace(" ", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
```

```python
df.dtypes
```

```python
df["TotalCharges"].fillna(df["TotalCharges"].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.to_csv("../outputs/telco_customer_churn_cleaned.csv", index=False)
```

```python
print("="*50)
print("Data Cleaning Completed")
print("="*50)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\nCleaned dataset saved successfully.")
```
